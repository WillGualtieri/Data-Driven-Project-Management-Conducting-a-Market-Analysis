```python
# Import the necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# year_str
workout_df = pd.read_csv("data/workout.csv")
peak_index = workout_df["workout_worldwide"].idxmax()
peak_month = workout_df.loc[peak_index, "month"]
year_str = peak_month.split("-")[0]
print(year_str)
```

```python
# peak_covid and current chart
three_keywords_df = pd.read_csv("data/three_keywords.csv")

import pandas as pd
import matplotlib.pyplot as plt

three_keywords_df = pd.read_csv("data/three_keywords.csv")
three_keywords_df["month"] = pd.to_datetime(three_keywords_df["month"])

keywords = {
    "home_workout_worldwide": "home workout",
    "gym_workout_worldwide": "gym workout",
    "home_gym_worldwide": "home gym",
}

fig, ax = plt.subplots(figsize=(12, 5))

for column, label in keywords.items():
    ax.plot(
        three_keywords_df["month"],
        three_keywords_df[column],
        label=label,
    )

ax.set(
    title="Worldwide workout search interest",
    xlabel="Month",
    ylabel="Search interest (0–100)",
)
ax.legend()
ax.grid(alpha=0.3)

fig.autofmt_xdate()
plt.tight_layout()
plt.show()
```

```python
# peak_covid
three_keywords_df_covid = three_keywords_df[(three_keywords_df["month"] > '2020-01') & (three_keywords_df
["month"] < '2020-12')]
covid_peaks = three_keywords_df_covid[list(keywords)].max()
peak_covid = keywords[covid_peaks.idxmax()]
print(peak_covid)
```

```python
# current
keyword_columns = [
    "home_workout_worldwide",
    "gym_workout_worldwide",
    "home_gym_worldwide",
]

latest_interest = (
    three_keywords_df
    .sort_values("month")[keyword_columns]
    .iloc[-1]
)

current = latest_interest.idxmax()

print(current)
```

```python
# top_country
workout_geo_df = pd.read_csv("data/workout_geo.csv")
countries = workout_geo_df[(workout_geo_df["country"] == 'United States') | (workout_geo_df["country"] == 'Austalia') | (workout_geo_df["country"] == 'Japan')]
top = countries["workout_2018_2023"].idxmax()
top_country = countries.loc[top, "country"]
print(top_country)
```
