```python
# Import the necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
workout_df = pd.read_csv("data/workout.csv")
year_str = workout_df[workout_df["workout_worldwide"] == 100]
year_str = year_str["month"].str.split('-').str.get(0)
# print(year_str)
```

```python
three_keywords_df = pd.read_csv("data/three_keywords.csv")
three_keywords_df_covid = three_keywords_df[three_keywords_df["month"] > '2020-01']
three_keywords_df_covid = three_keywords_df_covid[three_keywords_df_covid["month"] < '2020-12']
peak_covid1 = three_keywords_df_covid["home_workout_worldwide"].value_counts().idxmax()
peak_covid2 = three_keywords_df_covid["gym_workout_worldwide"].value_counts().idxmax()
peak_covid3 = three_keywords_df_covid["home_gym_worldwide"].value_counts().idxmax()
peak_covid_dict = {
    "home_workout_worldwide": peak_covid1,
    "gym_workout_worldwide": peak_covid2,
    "home_gym_worldwide": peak_covid3
}
peak_covid = max(peak_covid_dict.keys())
# print(peak_covid)
```

```python
three_keywords_df_current = three_keywords_df[three_keywords_df["month"] < '2023-03']
three_keywords_df_current = three_keywords_df_current[three_keywords_df_current["month"] > '2022-04']
current1 = three_keywords_df_current["home_workout_worldwide"].value_counts().idxmax()
current2 = three_keywords_df_current["gym_workout_worldwide"].value_counts().idxmax()
current3 = three_keywords_df_current["home_gym_worldwide"].value_counts().idxmax()
current_dict = {
    "home_workout_worldwide": current1,
    "gym_workout_worldwide": current2,
    "home_gym_worldwide": current3
}
current = max(current_dict.keys())
print(current)
```
